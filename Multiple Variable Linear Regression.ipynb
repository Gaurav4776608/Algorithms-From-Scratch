```python
x = [
    [1400, 3, 10, 5.2],
    [1800, 4, 5, 3.1],
    [1200, 2, 20, 8],
    [2200, 4, 2, 2.5],
    [1600, 3, 15, 6.4],
    [2800, 5, 8, 1.8],
    [1050, 2, 30, 10],
    [2400, 4, 3, 4],
    [1750, 3, 12, 7.2],
    [3100, 5, 1, 2]
]
```

```python
x
```

```python
#adding a column of 1 in x matrix
for i in range(0, len(x)):
  x[i].insert(0, 1)
x
```

```python
#transpose matrix

transpose_x = []
for i in range (0, len(x[0])):  #no of columns
  row = []
  for j in range (0, len(x)):   #no of rows
    row.append(x[j][i])
  transpose_x.append(row)
```

```python
y = [
    [245],
    [312],
    [178],
    [410],
    [268],
    [498],
    [145],
    [430],
    [290],
    [560]
]
```

```python
beta = [      #randomly taken numbers for initialisation
    [0.1],
    [0.3],
    [-0.2],
    [0.5],
    [0.8]
    ]
```

```python
beta
```

```python
y_cap = []   #predicted values

for i in range(0, len(x)):
  S = 0
  for j in range(0, len(x[i])):
    S += x[i][j] * beta[j][0]
  y_cap.append([S])

y_cap
```

```python
error = []   #error between actual and predicted values

for i in range(0, len(x)):
  error.append([y_cap[i][0]-y[i][0]])

error
```

```python
transpose_error = []
for i in range (0, len(error[0])):  #no of columns
  row = []
  for j in range (0, len(error)):   #no of rows
    row.append(error[j][i])
  transpose_error.append(row)

transpose_error
```

```python
m = len(error)  #number of rows
```

```python
square_of_error = []    # we are doing transpose of error * error for finding the cost function

S = 0

for i in range(0, len(error)):
  S += error[i][0] * transpose_error[0][i]

square_of_error.append([S])

square_of_error
```

```python
#cost function
J = 1/(2*m) * (square_of_error[0][0])
J
```

```python
gradient = []
for i in range(len(transpose_x)):
    S = 0
    for j in range(len(transpose_x[i])):
        S += transpose_x[i][j] * error[j][0]
    gradient.append([S])

m = len(y)
for i in range(0, len(gradient)):
    gradient[i][0] = gradient[i][0] / m

gradient
```

```python
alpha = 0.000000001
for i in range(0, len(beta)):
  beta[i][0] = beta[i][0] - (alpha*gradient[i][0])

beta
```

```python
# recalculate y_cap with new beta
y_cap = []
for i in range(len(x)):
    S = 0
    for j in range(len(x[i])):
        S += x[i][j] * beta[j][0]
    y_cap.append([S])

# recalculate error
error = []
for i in range(len(x)):
    error.append([y_cap[i][0] - y[i][0]])

# recalculate cost
S = 0
for i in range(len(error)):
    S += error[i][0] ** 2

J = S / (2 * m)
print(J)
```

```python
x = [
    [1400, 3, 10, 5.2],
    [1800, 4, 5, 3.1],
    [1200, 2, 20, 8],
    [2200, 4, 2, 2.5],
    [1600, 3, 15, 6.4],
    [2800, 5, 8, 1.8],
    [1050, 2, 30, 10],
    [2400, 4, 3, 4],
    [1750, 3, 12, 7.2],
    [3100, 5, 1, 2]
]

y = [
    [245],
    [312],
    [178],
    [410],
    [268],
    [498],
    [145],
    [430],
    [290],
    [560]
]

beta = [
    [0.1],
    [0.3],
    [-0.2],
    [0.5],
    [0.8]
]

num_features = len(x[0])
min_vals = []
max_vals = []

for j in range(num_features):
    col = [x[i][j] for i in range(len(x))]
    min_vals.append(min(col))
    max_vals.append(max(col))

x_norm = []
for i in range(len(x)):
    row = []
    for j in range(num_features):
        val = (x[i][j] - min_vals[j]) / (max_vals[j] - min_vals[j])
        row.append(val)
    x_norm.append(row)

for i in range(len(x_norm)):
    x_norm[i].insert(0, 1)

transpose_x = []
for i in range(len(x_norm[0])):
    row = []
    for j in range(len(x_norm)):
        row.append(x_norm[j][i])
    transpose_x.append(row)

epoch = 10000000
alpha = 0.01
m = len(y)

for i in range(epoch):

    y_cap = []
    for j in range(len(x_norm)):
        S = 0
        for k in range(len(x_norm[j])):
            S += x_norm[j][k] * beta[k][0]
        y_cap.append([S])

    error = []
    for j in range(len(y)):
        error.append([y_cap[j][0] - y[j][0]])

    S = 0
    for j in range(len(error)):
        S += error[j][0] ** 2
    cost = S / (2 * m)

    gradient = []
    for j in range(len(transpose_x)):
        S = 0
        for k in range(len(transpose_x[j])):
            S += transpose_x[j][k] * error[k][0]
        gradient.append([S / m])

    for j in range(len(beta)):
        beta[j][0] = beta[j][0] - (alpha * gradient[j][0])

    if i % 100 == 0:
        print(cost)
```
