```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('Student_Marks.csv')
x = df["time_study"].to_numpy()
y = df["Marks"].to_numpy()
epochs = 1000
m = 2
c = 10
alpha = 0.001

for i in range(0, epochs):
  predictions = []
  for j in range(0, x.size):
    predictions.append(m*x[j] + c)
  error = []
  for j in range(0, x.size):
    error.append((predictions[j]-y[j])**2)

  s = 0

  for j in range(0, x.size):
    s += error[j]

  cost_func = (1/(2*x.size)) * s
  print(cost_func)

  grad_m = 0
  grad_c = 0

  for j in range(0, x.size):
    grad_m += (predictions[j]-y[j]) * x[j]
    grad_c += (predictions[j]-y[j])

  grad_m = (1/x.size) * grad_m
  grad_c = (1/x.size) * grad_c

  m = m - (alpha*grad_m)
  c = c - (alpha*grad_c)
```
